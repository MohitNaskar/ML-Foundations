```python
from calendar import c

from sklearn.datasets import make_classification
import numpy as np
X,y = make_classification(n_samples=100,n_features=2,n_classes=2,n_informative=1,n_redundant=0,n_clusters_per_class=1,random_state=41,hypercube=False,class_sep=20)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='bwr')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Synthetic Classification Dataset')
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X,y)
print(f'Intercept: {model.intercept_}, Coefficients: {model.coef_}')
```

```python
def gd(X,y):
    X= np.insert(X,0,1,axis=1)
    w = np.ones(X.shape[1])
    lr = 0.1

    for _ in range(1000):
        y_hat = sigmoid(np.dot(X,w))
        w = w + lr*(np.dot(y-y_hat,X)/X.shape[0])
    return w[0], w[1:]  
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
gd(X,y)
```
